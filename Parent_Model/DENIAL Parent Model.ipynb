```python
global_rate = train_df["Decision"].mean()
```

```python
import pandas as pd

def smoothed_target_encoding(df, group_col, target_col, m=50):
    
    global_mean = df[target_col].mean()
    
    stats = df.groupby(group_col)[target_col].agg(["mean","count"])
    
    smooth = (stats["count"] * stats["mean"] + m * global_mean) / (stats["count"] + m)
    
    return smooth
```

```python
cpt_rate = smoothed_target_encoding(train_df, "CPT", "Decision")

train_df["cpt_rate"] = train_df["CPT"].map(cpt_rate)
test_df["cpt_rate"] = test_df["CPT"].map(cpt_rate)

train_df["cpt_rate"] = train_df["cpt_rate"].fillna(global_rate)
test_df["cpt_rate"] = test_df["cpt_rate"].fillna(global_rate)
```

```python
train_df["cpt_icd"] = train_df["CPT"] + "_" + train_df["D1"]
test_df["cpt_icd"] = test_df["CPT"] + "_" + test_df["D1"]

cpt_icd_rate = smoothed_target_encoding(train_df, "cpt_icd", "Decision")

train_df["cpt_icd_rate"] = train_df["cpt_icd"].map(cpt_icd_rate)
test_df["cpt_icd_rate"] = test_df["cpt_icd"].map(cpt_icd_rate)

train_df["cpt_icd_rate"] = train_df["cpt_icd_rate"].fillna(global_rate)
test_df["cpt_icd_rate"] = test_df["cpt_icd_rate"].fillna(global_rate)
```

```python
train_df["cpt_pair"] = train_df["CPT"] + "_" + train_df["other_cpt_1"]
test_df["cpt_pair"] = test_df["CPT"] + "_" + test_df["other_cpt_1"]

pair_rate = smoothed_target_encoding(train_df, "cpt_pair", "Decision")

train_df["cpt_pair_rate"] = train_df["cpt_pair"].map(pair_rate)
test_df["cpt_pair_rate"] = test_df["cpt_pair"].map(pair_rate)

train_df["cpt_pair_rate"] = train_df["cpt_pair_rate"].fillna(global_rate)
test_df["cpt_pair_rate"] = test_df["cpt_pair_rate"].fillna(global_rate)
```

```python
import pandas as pd
import pickle
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from catboost import CatBoostClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# ── 1. Load model & artefacts ────────────────────────────────────────────────
model = CatBoostClassifier()
model.load_model("catboost_denial_model.cbm")

lookup_tables = pickle.load(open("lookup_tables.pkl", "rb"))
feature_cols  = pickle.load(open("feature_columns.pkl", "rb"))
cat_cols      = pickle.load(open("cat_cols.pkl", "rb"))

# ── 2. Load raw data  ────────────────────────────────────────────────────────
ACCEPT_CSV = "acc_dec2025.csv"
DENY_CSV   = "den_dec2025.csv"

cols = [
    "CPT", "M1", "M2", "M3", "M4",
    "D1", "D2", "D3", "D4",
    "ABK",
    "ABF_1","ABF_2","ABF_3","ABF_4","ABF_5",
    "ABF_6","ABF_7","ABF_8","ABF_9","ABF_10","ABF_11",
    "patient_gender", "patient_age",
    "claim_no", "PAYER_ID", "POS", "Decision"
]

novacc = pd.read_csv(ACCEPT_CSV)[cols].drop_duplicates(subset=["claim_no", "CPT"])
novden = pd.read_csv(DENY_CSV)[cols].drop_duplicates(subset=["claim_no", "CPT"])
df_inf = pd.concat([novacc, novden], ignore_index=True)

# ── 3. Fill nulls ────────────────────────────────────────────────────────────
mod_cols = ["M1", "M2", "M3", "M4"]
dx_cols  = ["D1", "D2", "D3", "D4"]
abf_cols = ["ABF_1","ABF_2","ABF_3","ABF_4","ABF_5",
            "ABF_6","ABF_7","ABF_8","ABF_9","ABF_10","ABF_11"]

df_inf[mod_cols] = df_inf[mod_cols].fillna("NONE")
df_inf[dx_cols]  = df_inf[dx_cols].fillna("NONE")
df_inf[abf_cols] = df_inf[abf_cols].fillna("NONE")
df_inf["ABK"]    = df_inf["ABK"].fillna(df_inf["D1"])
df_inf["patient_gender"] = df_inf["patient_gender"].fillna("UNKNOWN")
df_inf["POS"] = df_inf["POS"].fillna("NONE")

code_cols = ["CPT"] + mod_cols + dx_cols + ["ABK"] + abf_cols
df_inf[code_cols] = df_inf[code_cols].astype(str)
df_inf["POS"] = df_inf["POS"].astype(str)

# ── 4. CPT co-occurrence features ────────────────────────────────────────────
df_inf["cpt_count"] = df_inf.groupby("claim_no")["CPT"].transform("count")
df_inf = df_inf.sort_values(["claim_no", "CPT"])

for i, shift in enumerate([-1, -2, -3], 1):
    df_inf[f"cpt_next{i}"] = df_inf.groupby("claim_no")["CPT"].shift(shift)
    df_inf[f"cpt_prev{i}"] = df_inf.groupby("claim_no")["CPT"].shift(-shift)

df_inf["other_cpt_1"] = df_inf["cpt_next1"].fillna(df_inf["cpt_prev1"])
df_inf["other_cpt_2"] = df_inf["cpt_next2"].fillna(df_inf["cpt_prev2"])
df_inf["other_cpt_3"] = df_inf["cpt_next3"].fillna(df_inf["cpt_prev3"])
df_inf[["other_cpt_1","other_cpt_2","other_cpt_3"]] = \
    df_inf[["other_cpt_1","other_cpt_2","other_cpt_3"]].fillna("NONE")

df_inf = df_inf.drop(columns=[f"cpt_next{i}" for i in range(1,4)] +
                              [f"cpt_prev{i}" for i in range(1,4)])

# ── 5. Lookup probability features ──────────────────────────────────────────
lk = lookup_tables
df_inf["cpt_rate"]      = df_inf["CPT"].map(lk["cpt_rate"]).fillna(lk["global_rate"])
df_inf["cpt_m1"]        = df_inf["CPT"] + "_" + df_inf["M1"]
df_inf["cpt_m1_rate"]   = df_inf["cpt_m1"].map(lk["cpt_m1_rate"]).fillna(lk["global_rate"])
df_inf["cpt_icd"]       = df_inf["CPT"] + "_" + df_inf["D1"]
df_inf["cpt_icd_rate"]  = df_inf["cpt_icd"].map(lk["cpt_icd_rate"]).fillna(lk["global_rate"])
df_inf["cpt_pair"]      = df_inf["CPT"] + "_" + df_inf["other_cpt_1"]
df_inf["cpt_pair_rate"] = df_inf["cpt_pair"].map(lk["pair_rate"]).fillna(lk["global_rate"])

# ── 6. Build X / y ───────────────────────────────────────────────────────────
y_true = df_inf["Decision"]
X_inf  = df_inf.drop(columns=["Decision", "claim_no", "cpt_m1", "cpt_icd", "cpt_pair"])
X_inf  = X_inf[feature_cols]
X_inf[cat_cols] = X_inf[cat_cols].fillna("NONE").astype(str)

# ── 7. Predict ───────────────────────────────────────────────────────────────
# Decision: 0 = Denied (positive class), 1 = Accepted (negative class)
probs_inf = model.predict_proba(X_inf)[:, 1]
preds_inf = (probs_inf >= 0.5).astype(int)

CONFIDENCE_THRESHOLD = 0.90
preds_90 = (probs_inf >= CONFIDENCE_THRESHOLD).astype(int)

CLASSES = [0, 1]

# Layout (0 is positive): rows=Actual, cols=Predicted, order [0, 1]
#           Pred 0   Pred 1
#  Actual 0   TP       FN
#  Actual 1   FP       TN
CELL_TAG = np.array([["TP", "FN"],
                     ["FP", "TN"]])

def print_cm_summary(cm, total):
    # cm.ravel() = [cm[0,0], cm[0,1], cm[1,0], cm[1,1]] = [TP, FN, FP, TN]
    tp, fn, fp, tn = cm.ravel()
    print(f"  {'Metric':<32} {'Count':>8}   {'%':>7}")
    print(f"  {'-'*52}")
    print(f"  {'TP  — Actual 0 → Pred 0':<32} {tp:>8,}   {tp/total*100:>6.1f}%")
    print(f"  {'FN  — Actual 0 → Pred 1':<32} {fn:>8,}   {fn/total*100:>6.1f}%")
    print(f"  {'FP  — Actual 1 → Pred 0':<32} {fp:>8,}   {fp/total*100:>6.1f}%")
    print(f"  {'TN  — Actual 1 → Pred 1':<32} {tn:>8,}   {tn/total*100:>6.1f}%")
    print(f"  {'-'*52}")
    print(f"  {'Total':<32} {total:>8,}   100.0%\n")

def plot_cm(ax, cm, title, cmap):
    total = cm.sum()
    annot = np.array([[f"{CELL_TAG[r,c]}\n{cm[r,c]:,}\n({cm[r,c]/total*100:.1f}%)"
                       for c in range(2)] for r in range(2)])
    df_cm = pd.DataFrame(cm,
                         index=pd.Index([0, 1], name="Actual"),
                         columns=pd.Index([0, 1], name="Predicted"))
    sns.heatmap(df_cm, annot=annot, fmt="", cmap=cmap, ax=ax,
                linewidths=0.5, linecolor="gray", annot_kws={"size": 11})
    ax.set_title(title, fontsize=12)

total = len(y_true)

# ── 8. Threshold 0.50 ────────────────────────────────────────────────────────
print("=" * 55)
print("  Threshold = 0.50  (default)")
print("=" * 55)
print(f"Accuracy: {accuracy_score(y_true, preds_inf):.4f}\n")
cm_50 = confusion_matrix(y_true, preds_inf, labels=CLASSES)
print_cm_summary(cm_50, total)
print(classification_report(y_true, preds_inf, labels=CLASSES))

# ── 9. Threshold 0.90 ────────────────────────────────────────────────────────
print("=" * 55)
print(f"  Threshold = {CONFIDENCE_THRESHOLD}  (90% confidence)")
print("=" * 55)
high_conf_mask = (probs_inf >= CONFIDENCE_THRESHOLD) | (probs_inf <= (1 - CONFIDENCE_THRESHOLD))
n_covered = int(high_conf_mask.sum())
print(f"Rows covered at ≥90% confidence: {n_covered:,} / {total:,} ({n_covered/total*100:.1f}%)\n")
cm_90 = confusion_matrix(y_true, preds_90, labels=CLASSES)
print_cm_summary(cm_90, total)
print(classification_report(y_true, preds_90, labels=CLASSES))

# ── 10. Side-by-side confusion matrices ──────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
plot_cm(axes[0], cm_50, "Confusion Matrix — Threshold 0.50",  cmap="Blues")
plot_cm(axes[1], cm_90, f"Confusion Matrix — Threshold {CONFIDENCE_THRESHOLD}", cmap="Oranges")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import pickle
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from catboost import CatBoostClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# ── 1. Load model & artefacts ────────────────────────────────────────────────
model = CatBoostClassifier()
model.load_model("catboost_denial_model.cbm")

lookup_tables = pickle.load(open("lookup_tables.pkl", "rb"))
feature_cols  = pickle.load(open("feature_columns.pkl", "rb"))
cat_cols      = pickle.load(open("cat_cols.pkl", "rb"))

# ── 2. Load raw data  ────────────────────────────────────────────────────────
ACCEPT_CSV = "acc_feb2026.csv"
DENY_CSV   = "den_feb2026.csv"

cols = [
    "CPT", "M1", "M2", "M3", "M4",
    "D1", "D2", "D3", "D4",
    "ABK",
    "ABF_1","ABF_2","ABF_3","ABF_4","ABF_5",
    "ABF_6","ABF_7","ABF_8","ABF_9","ABF_10","ABF_11",
    "patient_gender", "patient_age",
    "claim_no", "PAYER_ID", "POS", "Decision"
]

novacc = pd.read_csv(ACCEPT_CSV)[cols].drop_duplicates(subset=["claim_no", "CPT"])
novden = pd.read_csv(DENY_CSV)[cols].drop_duplicates(subset=["claim_no", "CPT"])
df_inf = pd.concat([novacc, novden], ignore_index=True)

# ── 3. Fill nulls ────────────────────────────────────────────────────────────
mod_cols = ["M1", "M2", "M3", "M4"]
dx_cols  = ["D1", "D2", "D3", "D4"]
abf_cols = ["ABF_1","ABF_2","ABF_3","ABF_4","ABF_5",
            "ABF_6","ABF_7","ABF_8","ABF_9","ABF_10","ABF_11"]

df_inf[mod_cols] = df_inf[mod_cols].fillna("NONE")
df_inf[dx_cols]  = df_inf[dx_cols].fillna("NONE")
df_inf[abf_cols] = df_inf[abf_cols].fillna("NONE")
df_inf["ABK"]    = df_inf["ABK"].fillna(df_inf["D1"])
df_inf["patient_gender"] = df_inf["patient_gender"].fillna("UNKNOWN")
df_inf["POS"] = df_inf["POS"].fillna("NONE")

code_cols = ["CPT"] + mod_cols + dx_cols + ["ABK"] + abf_cols
df_inf[code_cols] = df_inf[code_cols].astype(str)
df_inf["POS"] = df_inf["POS"].astype(str)

# ── 4. CPT co-occurrence features ────────────────────────────────────────────
df_inf["cpt_count"] = df_inf.groupby("claim_no")["CPT"].transform("count")
df_inf = df_inf.sort_values(["claim_no", "CPT"])

for i, shift in enumerate([-1, -2, -3], 1):
    df_inf[f"cpt_next{i}"] = df_inf.groupby("claim_no")["CPT"].shift(shift)
    df_inf[f"cpt_prev{i}"] = df_inf.groupby("claim_no")["CPT"].shift(-shift)

df_inf["other_cpt_1"] = df_inf["cpt_next1"].fillna(df_inf["cpt_prev1"])
df_inf["other_cpt_2"] = df_inf["cpt_next2"].fillna(df_inf["cpt_prev2"])
df_inf["other_cpt_3"] = df_inf["cpt_next3"].fillna(df_inf["cpt_prev3"])
df_inf[["other_cpt_1","other_cpt_2","other_cpt_3"]] = \
    df_inf[["other_cpt_1","other_cpt_2","other_cpt_3"]].fillna("NONE")

df_inf = df_inf.drop(columns=[f"cpt_next{i}" for i in range(1,4)] +
                              [f"cpt_prev{i}" for i in range(1,4)])

# ── 5. Lookup probability features ──────────────────────────────────────────
lk = lookup_tables
df_inf["cpt_rate"]      = df_inf["CPT"].map(lk["cpt_rate"]).fillna(lk["global_rate"])
df_inf["cpt_m1"]        = df_inf["CPT"] + "_" + df_inf["M1"]
df_inf["cpt_m1_rate"]   = df_inf["cpt_m1"].map(lk["cpt_m1_rate"]).fillna(lk["global_rate"])
df_inf["cpt_icd"]       = df_inf["CPT"] + "_" + df_inf["D1"]
df_inf["cpt_icd_rate"]  = df_inf["cpt_icd"].map(lk["cpt_icd_rate"]).fillna(lk["global_rate"])
df_inf["cpt_pair"]      = df_inf["CPT"] + "_" + df_inf["other_cpt_1"]
df_inf["cpt_pair_rate"] = df_inf["cpt_pair"].map(lk["pair_rate"]).fillna(lk["global_rate"])

# ── 6. Build X / y ───────────────────────────────────────────────────────────
y_true = df_inf["Decision"]
X_inf  = df_inf.drop(columns=["Decision", "claim_no", "cpt_m1", "cpt_icd", "cpt_pair"])
X_inf  = X_inf[feature_cols]
X_inf[cat_cols] = X_inf[cat_cols].fillna("NONE").astype(str)

# ── 7. Predict ───────────────────────────────────────────────────────────────
# Decision: 0 = Denied (positive class), 1 = Accepted (negative class)
probs_inf = model.predict_proba(X_inf)[:, 1]
preds_inf = (probs_inf >= 0.5).astype(int)

CONFIDENCE_THRESHOLD = 0.90
preds_90 = (probs_inf >= CONFIDENCE_THRESHOLD).astype(int)

CLASSES = [0, 1]

# Layout (0 is positive): rows=Actual, cols=Predicted, order [0, 1]
#           Pred 0   Pred 1
#  Actual 0   TP       FN
#  Actual 1   FP       TN
CELL_TAG = np.array([["TP", "FN"],
                     ["FP", "TN"]])

def print_cm_summary(cm, total):
    # cm.ravel() = [cm[0,0], cm[0,1], cm[1,0], cm[1,1]] = [TP, FN, FP, TN]
    tp, fn, fp, tn = cm.ravel()
    print(f"  {'Metric':<32} {'Count':>8}   {'%':>7}")
    print(f"  {'-'*52}")
    print(f"  {'TP  — Actual 0 → Pred 0':<32} {tp:>8,}   {tp/total*100:>6.1f}%")
    print(f"  {'FN  — Actual 0 → Pred 1':<32} {fn:>8,}   {fn/total*100:>6.1f}%")
    print(f"  {'FP  — Actual 1 → Pred 0':<32} {fp:>8,}   {fp/total*100:>6.1f}%")
    print(f"  {'TN  — Actual 1 → Pred 1':<32} {tn:>8,}   {tn/total*100:>6.1f}%")
    print(f"  {'-'*52}")
    print(f"  {'Total':<32} {total:>8,}   100.0%\n")

def plot_cm(ax, cm, title, cmap):
    total = cm.sum()
    annot = np.array([[f"{CELL_TAG[r,c]}\n{cm[r,c]:,}\n({cm[r,c]/total*100:.1f}%)"
                       for c in range(2)] for r in range(2)])
    df_cm = pd.DataFrame(cm,
                         index=pd.Index([0, 1], name="Actual"),
                         columns=pd.Index([0, 1], name="Predicted"))
    sns.heatmap(df_cm, annot=annot, fmt="", cmap=cmap, ax=ax,
                linewidths=0.5, linecolor="gray", annot_kws={"size": 11})
    ax.set_title(title, fontsize=12)

total = len(y_true)

# ── 8. Threshold 0.50 ────────────────────────────────────────────────────────
print("=" * 55)
print("  Threshold = 0.50  (default)")
print("=" * 55)
print(f"Accuracy: {accuracy_score(y_true, preds_inf):.4f}\n")
cm_50 = confusion_matrix(y_true, preds_inf, labels=CLASSES)
print_cm_summary(cm_50, total)
print(classification_report(y_true, preds_inf, labels=CLASSES))

# ── 9. Threshold 0.90 ────────────────────────────────────────────────────────
print("=" * 55)
print(f"  Threshold = {CONFIDENCE_THRESHOLD}  (90% confidence)")
print("=" * 55)
high_conf_mask = (probs_inf >= CONFIDENCE_THRESHOLD) | (probs_inf <= (1 - CONFIDENCE_THRESHOLD))
n_covered = int(high_conf_mask.sum())
print(f"Rows covered at ≥90% confidence: {n_covered:,} / {total:,} ({n_covered/total*100:.1f}%)\n")
cm_90 = confusion_matrix(y_true, preds_90, labels=CLASSES)
print_cm_summary(cm_90, total)
print(classification_report(y_true, preds_90, labels=CLASSES))

# ── 10. Side-by-side confusion matrices ──────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
plot_cm(axes[0], cm_50, "Confusion Matrix — Threshold 0.50",  cmap="Blues")
plot_cm(axes[1], cm_90, f"Confusion Matrix — Threshold {CONFIDENCE_THRESHOLD}", cmap="Oranges")
plt.tight_layout()
plt.show()
```
