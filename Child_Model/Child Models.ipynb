```python

# ============================================================
# FULL PIPELINE — All steps combined
# ============================================================

import pandas as pd
from sklearn.model_selection import train_test_split

# ----------------------------------------------------------
# 1. Load data
# ----------------------------------------------------------
pd.set_option('display.max_columns', None)
jannovden = pd.read_csv("den_jan2025-nov2025.csv")

# ----------------------------------------------------------
# 2. Remove duplicates
# ----------------------------------------------------------
df = jannovden.drop_duplicates().reset_index(drop=True)

# ----------------------------------------------------------
# 3. Drop rows with missing CPT
# ----------------------------------------------------------
df = df.dropna(subset=["CPT"])

# ----------------------------------------------------------
# 4. Normalize target column
# ----------------------------------------------------------
target = "Rejection_Description"
df[target] = df[target].str.lower().str.strip()

# ----------------------------------------------------------
# 5. Remove exact-duplicate / previously-paid denials
# ----------------------------------------------------------
drop_reasons = [
    "duplicate",
    "duplicate/previously paid",
    "previously paid. payment for this claim/service may have been provided in a previous payment"
]
df = df[~df[target].isin(drop_reasons)]

# ----------------------------------------------------------
# 6. Map raw rejection descriptions to categories
# ----------------------------------------------------------
category_map = {
    "invalid modifier": "invalid_modifier",
    "lack of info": "lack_of_info",
    "ins need info from patient": "lack_of_info",
    "required medical notes": "lack_of_info",
    "non covered benefits": "noncovered_benefits",
    "benefits exhausted": "noncovered_benefits",
    "covered under capitation": "noncovered_benefits",
    "patient enrolled in a hospice": "noncovered_benefits",
    "inclusive in other procedure": "inclusive_procedure",
    "procedure code incidental to primary procedure": "inclusive_procedure",
    "cpt code is bundled with another billed cpt code on the same day": "inclusive_procedure",
    "this procedure is not paid separately": "inclusive_procedure",
    "mutually exclusive procedure": "inclusive_procedure",
    "incidental procedure": "inclusive_procedure",
    "cpt inconsistent with icd": "cpt_icd_mismatch",
    "the procedure or service is inconsistent with the patient s history": "cpt_icd_mismatch",
    "inc dx": "cpt_icd_mismatch",
    "invalid pos": "invalid_pos",
    "inconsistent with patient gender": "gender_inconsistent",
    "inconsistent with patient age": "age_inconsistent",
    "out of network provider": "payer_rule",
    "services not provided by network/primary care providers": "payer_rule",
    "non-participating provider": "payer_rule",
    "no out of network benefits": "payer_rule",
    "invalid/missing authorization no": "authorization",
    "invalid/missing referral no.": "authorization",
    "services not authorized by medipass pcp": "authorization",
    "referring provider number not on file": "authorization",
    "provider is not patient's pcp": "authorization",
    "service performed by another pcp within 60 days": "authorization",
    "not eligible charges": "eligibility",
    "subscriber/insured not found": "eligibility",
    "member not found": "eligibility",
    "provider was not eligible on dos": "eligibility",
    "ins termed": "eligibility",
}
df[target] = df[target].map(category_map)

# ----------------------------------------------------------
# 7. Resolve multiple denials per (claim_no, CPT) by priority
# ----------------------------------------------------------
priority = {
    "invalid_modifier": 1,
    "inclusive_procedure": 2,
    "lack_of_info": 3,
    "noncovered_benefits": 4,
    "cpt_icd_mismatch": 5,
    "invalid_pos": 6,
    "gender_inconsistent": 7,
    "age_inconsistent": 8,
    "authorization": 9,
    "eligibility": 10,
    "payer_rule": 11,
}
df["priority"] = df[target].map(priority)
df = (
    df.sort_values("priority")
      .drop_duplicates(subset=["claim_no", "CPT"], keep="first")
)

# Drop rows where category is NaN (unmapped) — labelled "cat" then filtered
df[target] = df[target].fillna("cat")
df = df[df[target] != "cat"]

# ----------------------------------------------------------
# 8. Subset to relevant columns
# ----------------------------------------------------------
feature_cols = [
    "CPT",
    "M1", "M2", "M3", "M4",
    "D1", "D2", "D3", "D4",
    "P1", "P2", "P3", "P4",
    "ABK",
    "ABF_1", "ABF_2", "ABF_3", "ABF_4", "ABF_5",
    "ABF_6", "ABF_7", "ABF_8", "ABF_9", "ABF_10", "ABF_11",
    "patient_gender",
    "patient_age",
    "PAYER_ID",
    "POS",
]
df = df[feature_cols + ["claim_no", target]].copy()

# ----------------------------------------------------------
# 9. Fill missing code columns with "MISSING"
# ----------------------------------------------------------
code_cols = [
    "CPT",
    "M1", "M2", "M3", "M4",
    "D1", "D2", "D3", "D4",
    "P1", "P2", "P3", "P4",
    "ABK",
    "ABF_1", "ABF_2", "ABF_3", "ABF_4", "ABF_5",
    "ABF_6", "ABF_7", "ABF_8", "ABF_9", "ABF_10", "ABF_11",
    "POS",
]
df[code_cols] = df[code_cols].fillna("MISSING")

# ----------------------------------------------------------
# 10. Feature engineering
# ----------------------------------------------------------
# Other CPTs billed on the same claim
df = df.sort_values(["claim_no", "CPT"])
claim_cpts = df.groupby("claim_no")["CPT"].apply(list)
df = df.merge(claim_cpts.rename("all_cpts"), on="claim_no")
df["other_cpts"] = df.apply(lambda x: [c for c in x["all_cpts"] if c != x["CPT"]], axis=1)
df["other_cpt_1"] = df["other_cpts"].str[0]
df["other_cpt_2"] = df["other_cpts"].str[1]
df["other_cpt_3"] = df["other_cpts"].str[2]
df[["other_cpt_1", "other_cpt_2", "other_cpt_3"]] = df[["other_cpt_1", "other_cpt_2", "other_cpt_3"]].fillna("NONE")
df = df.drop(columns=["all_cpts", "other_cpts"])

# Count features
mod_cols = ["M1", "M2", "M3", "M4"]
diag_cols = ["D1", "D2", "D3", "D4", "ABK",
             "ABF_1", "ABF_2", "ABF_3", "ABF_4", "ABF_5",
             "ABF_6", "ABF_7", "ABF_8", "ABF_9", "ABF_10", "ABF_11"]

df["modifier_count"]     = (df[mod_cols] != "MISSING").sum(axis=1)
df["diagnosis_count"]    = (df[diag_cols] != "MISSING").sum(axis=1)
df["cpt_count"]          = df.groupby("claim_no")["CPT"].transform("count")
df["unique_cpt_count"]   = df.groupby("claim_no")["CPT"].transform("nunique")

feature_cols += ["modifier_count", "diagnosis_count", "cpt_count", "unique_cpt_count"]

# ----------------------------------------------------------
# 11. Claim-level stratified train / test split
# ----------------------------------------------------------
unique_claims = (
    df.groupby("claim_no")[target]
      .first()
      .reset_index()
)

train_claims, test_claims = train_test_split(
    unique_claims,
    test_size=0.2,
    stratify=unique_claims[target],
    random_state=42,
)

train_df = df[df["claim_no"].isin(train_claims["claim_no"])].copy()
test_df  = df[df["claim_no"].isin(test_claims["claim_no"])].copy()

print("Train size:", len(train_df), "| Test size:", len(test_df))
print("Overlap (should be 0):", len(set(train_df["claim_no"]) & set(test_df["claim_no"])))

# ----------------------------------------------------------
# 12. Probability encoding (target encoding)
# ----------------------------------------------------------
def probability_encoder(train_df, test_df, col, target, lookup_tables):
    prob = (
        train_df
        .groupby(col)[target]
        .value_counts(normalize=True)
        .unstack()
    )
    lookup_tables[col] = prob
    train_df = train_df.join(prob, on=col, rsuffix=f"_{col}_prob")
    test_df  = test_df.join(prob,  on=col, rsuffix=f"_{col}_prob")
    rename_map = {c: f"{col}_prob_{c}" for c in prob.columns}
    train_df.rename(columns=rename_map, inplace=True)
    test_df.rename(columns=rename_map,  inplace=True)
    return train_df, test_df
```

```python
import pickle

# ------------------------------------------------------------------ #
#  Probability encoding  (NO leakage — all tables built on train only)
# ------------------------------------------------------------------ #

target = "Target"

prob_cols = [
    "CPT",
    "M1", "M2", "M3", "M4",
    "D1", "D2", "D3", "D4",
    "ABK",
    "PAYER_ID",
    "POS",
]

# Step 1 – cast to string in both splits BEFORE any encoding
for col in prob_cols:
    train_df[col] = train_df[col].astype(str)
    test_df[col]  = test_df[col].astype(str)

# Step 2 – build lookup tables from train ONLY, then map to both splits
lookup_tables = {}

for col in prob_cols:
    print(f"Encoding: {col}")

    # P(class | col_value)  — computed on train only
    prob = (
        train_df
        .groupby(col)[target]
        .value_counts(normalize=True)
        .unstack(fill_value=0)
    )

    lookup_tables[col] = prob

    # Map each class probability as a new column in both sets.
    # Unseen test categories → NaN → filled with 0  (conservative: no info = 0 signal)
    for cls in prob.columns:
        new_col = f"{col}_prob_{cls}"
        mapping  = prob[cls]
        train_df[new_col] = train_df[col].map(mapping).fillna(0)
        test_df[new_col]  = test_df[col].map(mapping).fillna(0)

# Step 3 – persist lookup tables for inference time
pickle.dump(lookup_tables, open("probability_lookup_tables.pkl", "wb"))

print("\nDone.")
print(f"Lookup tables saved → probability_lookup_tables.pkl")
print(f"train_df shape : {train_df.shape}")
print(f"test_df  shape : {test_df.shape}")

# Quick sanity check – verify no leakage
assert len(set(train_df["claim_no"]) & set(test_df["claim_no"])) == 0, \
    "LEAKAGE DETECTED: claim_no overlap between train and test!"
print("Leakage check passed ✓  (no claim_no overlap)")
```

```python
import numpy as np
from catboost import CatBoostClassifier
from sklearn.metrics import classification_report, accuracy_score
from sklearn.preprocessing import LabelEncoder

# ------------------------------------------------------------------ #
#  1. Drop claim_no — was only needed for leakage-safe splitting
# ------------------------------------------------------------------ #
train_df = train_df.drop(columns=["claim_no"], errors="ignore")
test_df  = test_df.drop(columns=["claim_no"],  errors="ignore")

# ------------------------------------------------------------------ #
#  2. Build X / y — exclude original categorical cols & target
#     Keep only numeric probability-encoded columns + numeric features
# ------------------------------------------------------------------ #
non_feature_cols = prob_cols + [target, "priority"]

X_train = train_df.drop(columns=non_feature_cols, errors="ignore").select_dtypes(include=[np.number])
y_train = train_df[target]

X_test  = test_df.drop(columns=non_feature_cols, errors="ignore").select_dtypes(include=[np.number])
y_test  = test_df[target]

print("X_train shape:", X_train.shape)
print("X_test  shape:", X_test.shape)
print("Classes       :", sorted(y_train.unique()))

# ------------------------------------------------------------------ #
#  3. Compute inverse-frequency class weights
# ------------------------------------------------------------------ #
class_counts = y_train.value_counts()
total        = len(y_train)
n_classes    = len(class_counts)

# weight = total / (n_classes * count)  — standard balanced weighting
class_weights = {cls: total / (n_classes * cnt) for cls, cnt in class_counts.items()}
print("\nClass weights:")
for cls, w in sorted(class_weights.items()):
    print(f"  {cls:<30s}  {w:.4f}")

# ------------------------------------------------------------------ #
#  4. Train CatBoost on GPU with class weights
# ------------------------------------------------------------------ #
model = CatBoostClassifier(
    iterations       = 5000,
    learning_rate    = 0.05,
    depth            = 6,
    loss_function    = "MultiClass",
    eval_metric      = "Accuracy",
    class_weights    = class_weights,
    task_type        = "GPU",
    devices          = "0",
    random_seed      = 42,
    verbose          = 100,
    early_stopping_rounds = 50,
)

model.fit(
    X_train, y_train,
    eval_set=(X_test, y_test),
    use_best_model=True,
)

# ------------------------------------------------------------------ #
#  5. Evaluate — overall + per-class accuracy
# ------------------------------------------------------------------ #
y_pred = model.predict(X_test).flatten()

print("\n" + "="*60)
print(f"Overall Test Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print("="*60)

print("\nPer-class report:\n")
print(classification_report(y_test, y_pred, digits=4))

# Per-class accuracy explicitly
print("Per-class accuracy:")
classes = sorted(y_test.unique())
for cls in classes:
    mask    = y_test == cls
    cls_acc = accuracy_score(y_test[mask], y_pred[mask])
    support = mask.sum()
    print(f"  {cls:<30s}  acc={cls_acc:.4f}  support={support}")

# ------------------------------------------------------------------ #
#  6. Save model
# ------------------------------------------------------------------ #
model.save_model("catboost_denial_multiclass.cbm")
print("\nModel saved → catboost_denial_multiclass.cbm")
```

```python

# ============================================================
# FORWARD TEST — den_jan2026.csv  (January 2026)
# CatBoost only. Loads saved lookup tables + model for pure inference.
# ============================================================
import pandas as pd
import numpy as np
import pickle
from sklearn.metrics import accuracy_score, classification_report
from catboost import CatBoostClassifier

# ----------------------------------------------------------
# Hardcoded training feature columns (exact order from X_train)
# ----------------------------------------------------------
TRAIN_COLS = [
    "P1", "patient_age", "cpt_count", "modifier_count", "diagnosis_count", "unique_cpt_count",
    "CPT_prob_age_inconsistent", "CPT_prob_authorization", "CPT_prob_cpt_icd_mismatch",
    "CPT_prob_eligibility", "CPT_prob_gender_inconsistent", "CPT_prob_inclusive_procedure",
    "CPT_prob_invalid_modifier", "CPT_prob_invalid_pos", "CPT_prob_lack_of_info",
    "CPT_prob_noncovered_benefits", "CPT_prob_payer_rule",
    "other_cpt_1_prob_age_inconsistent", "other_cpt_1_prob_authorization", "other_cpt_1_prob_cpt_icd_mismatch",
    "other_cpt_1_prob_eligibility", "other_cpt_1_prob_gender_inconsistent", "other_cpt_1_prob_inclusive_procedure",
    "other_cpt_1_prob_invalid_modifier", "other_cpt_1_prob_invalid_pos", "other_cpt_1_prob_lack_of_info",
    "other_cpt_1_prob_noncovered_benefits", "other_cpt_1_prob_payer_rule",
    "other_cpt_2_prob_age_inconsistent", "other_cpt_2_prob_authorization", "other_cpt_2_prob_cpt_icd_mismatch",
    "other_cpt_2_prob_eligibility", "other_cpt_2_prob_gender_inconsistent", "other_cpt_2_prob_inclusive_procedure",
    "other_cpt_2_prob_invalid_modifier", "other_cpt_2_prob_invalid_pos", "other_cpt_2_prob_lack_of_info",
    "other_cpt_2_prob_noncovered_benefits", "other_cpt_2_prob_payer_rule",
    "other_cpt_3_prob_age_inconsistent", "other_cpt_3_prob_authorization", "other_cpt_3_prob_cpt_icd_mismatch",
    "other_cpt_3_prob_eligibility", "other_cpt_3_prob_gender_inconsistent", "other_cpt_3_prob_inclusive_procedure",
    "other_cpt_3_prob_invalid_modifier", "other_cpt_3_prob_invalid_pos", "other_cpt_3_prob_lack_of_info",
    "other_cpt_3_prob_noncovered_benefits", "other_cpt_3_prob_payer_rule",
    "M1_prob_age_inconsistent", "M1_prob_authorization", "M1_prob_cpt_icd_mismatch",
    "M1_prob_eligibility", "M1_prob_gender_inconsistent", "M1_prob_inclusive_procedure",
    "M1_prob_invalid_modifier", "M1_prob_invalid_pos", "M1_prob_lack_of_info",
    "M1_prob_noncovered_benefits", "M1_prob_payer_rule",
    "M2_prob_age_inconsistent", "M2_prob_authorization", "M2_prob_cpt_icd_mismatch",
    "M2_prob_eligibility", "M2_prob_gender_inconsistent", "M2_prob_inclusive_procedure",
    "M2_prob_invalid_modifier", "M2_prob_invalid_pos", "M2_prob_lack_of_info",
    "M2_prob_noncovered_benefits", "M2_prob_payer_rule",
    "M3_prob_age_inconsistent", "M3_prob_authorization", "M3_prob_cpt_icd_mismatch",
    "M3_prob_eligibility", "M3_prob_gender_inconsistent", "M3_prob_inclusive_procedure",
    "M3_prob_invalid_modifier", "M3_prob_invalid_pos", "M3_prob_lack_of_info",
    "M3_prob_noncovered_benefits", "M3_prob_payer_rule",
    "M4_prob_age_inconsistent", "M4_prob_authorization", "M4_prob_cpt_icd_mismatch",
    "M4_prob_eligibility", "M4_prob_gender_inconsistent", "M4_prob_inclusive_procedure",
    "M4_prob_invalid_modifier", "M4_prob_invalid_pos", "M4_prob_lack_of_info",
    "M4_prob_noncovered_benefits", "M4_prob_payer_rule",
    "D1_prob_age_inconsistent", "D1_prob_authorization", "D1_prob_cpt_icd_mismatch",
    "D1_prob_eligibility", "D1_prob_gender_inconsistent", "D1_prob_inclusive_procedure",
    "D1_prob_invalid_modifier", "D1_prob_invalid_pos", "D1_prob_lack_of_info",
    "D1_prob_noncovered_benefits", "D1_prob_payer_rule",
    "D2_prob_age_inconsistent", "D2_prob_authorization", "D2_prob_cpt_icd_mismatch",
    "D2_prob_eligibility", "D2_prob_gender_inconsistent", "D2_prob_inclusive_procedure",
    "D2_prob_invalid_modifier", "D2_prob_invalid_pos", "D2_prob_lack_of_info",
    "D2_prob_noncovered_benefits", "D2_prob_payer_rule",
    "D3_prob_age_inconsistent", "D3_prob_authorization", "D3_prob_cpt_icd_mismatch",
    "D3_prob_eligibility", "D3_prob_gender_inconsistent", "D3_prob_inclusive_procedure",
    "D3_prob_invalid_modifier", "D3_prob_invalid_pos", "D3_prob_lack_of_info",
    "D3_prob_noncovered_benefits", "D3_prob_payer_rule",
    "D4_prob_age_inconsistent", "D4_prob_authorization", "D4_prob_cpt_icd_mismatch",
    "D4_prob_eligibility", "D4_prob_gender_inconsistent", "D4_prob_inclusive_procedure",
    "D4_prob_invalid_modifier", "D4_prob_invalid_pos", "D4_prob_lack_of_info",
    "D4_prob_noncovered_benefits", "D4_prob_payer_rule",
    "ABK_prob_age_inconsistent", "ABK_prob_authorization", "ABK_prob_cpt_icd_mismatch",
    "ABK_prob_eligibility", "ABK_prob_gender_inconsistent", "ABK_prob_inclusive_procedure",
    "ABK_prob_invalid_modifier", "ABK_prob_invalid_pos", "ABK_prob_lack_of_info",
    "ABK_prob_noncovered_benefits", "ABK_prob_payer_rule",
    "PAYER_ID_prob_age_inconsistent", "PAYER_ID_prob_authorization", "PAYER_ID_prob_cpt_icd_mismatch",
    "PAYER_ID_prob_eligibility", "PAYER_ID_prob_gender_inconsistent", "PAYER_ID_prob_inclusive_procedure",
    "PAYER_ID_prob_invalid_modifier", "PAYER_ID_prob_invalid_pos", "PAYER_ID_prob_lack_of_info",
    "PAYER_ID_prob_noncovered_benefits", "PAYER_ID_prob_payer_rule",
    "POS_prob_age_inconsistent", "POS_prob_authorization", "POS_prob_cpt_icd_mismatch",
    "POS_prob_eligibility", "POS_prob_gender_inconsistent", "POS_prob_inclusive_procedure",
    "POS_prob_invalid_modifier", "POS_prob_invalid_pos", "POS_prob_lack_of_info",
    "POS_prob_noncovered_benefits", "POS_prob_payer_rule",
]  # 171 features

# ----------------------------------------------------------
# 1. Load new data
# ----------------------------------------------------------
fwd = pd.read_csv("den_jan2026.csv")
pd.set_option('display.max_columns', None)
print("Raw shape:", fwd.shape)

# ----------------------------------------------------------
# 2. Preprocessing (identical to training pipeline)
# ----------------------------------------------------------
fwd = fwd.drop_duplicates().reset_index(drop=True)
fwd = fwd.dropna(subset=["CPT"])

target = "Rejection_Description"
fwd[target] = fwd[target].str.lower().str.strip()

drop_reasons = [
    "duplicate",
    "duplicate/previously paid",
    "previously paid. payment for this claim/service may have been provided in a previous payment",
]
fwd = fwd[~fwd[target].isin(drop_reasons)]

category_map = {
    "invalid modifier": "invalid_modifier",
    "lack of info": "lack_of_info",
    "ins need info from patient": "lack_of_info",
    "required medical notes": "lack_of_info",
    "non covered benefits": "noncovered_benefits",
    "benefits exhausted": "noncovered_benefits",
    "covered under capitation": "noncovered_benefits",
    "patient enrolled in a hospice": "noncovered_benefits",
    "inclusive in other procedure": "inclusive_procedure",
    "procedure code incidental to primary procedure": "inclusive_procedure",
    "cpt code is bundled with another billed cpt code on the same day": "inclusive_procedure",
    "this procedure is not paid separately": "inclusive_procedure",
    "mutually exclusive procedure": "inclusive_procedure",
    "incidental procedure": "inclusive_procedure",
    "cpt inconsistent with icd": "cpt_icd_mismatch",
    "the procedure or service is inconsistent with the patient s history": "cpt_icd_mismatch",
    "inc dx": "cpt_icd_mismatch",
    "invalid pos": "invalid_pos",
    "inconsistent with patient gender": "gender_inconsistent",
    "inconsistent with patient age": "age_inconsistent",
    "out of network provider": "payer_rule",
    "services not provided by network/primary care providers": "payer_rule",
    "non-participating provider": "payer_rule",
    "no out of network benefits": "payer_rule",
    "invalid/missing authorization no": "authorization",
    "invalid/missing referral no.": "authorization",
    "services not authorized by medipass pcp": "authorization",
    "referring provider number not on file": "authorization",
    "provider is not patient's pcp": "authorization",
    "service performed by another pcp within 60 days": "authorization",
    "not eligible charges": "eligibility",
    "subscriber/insured not found": "eligibility",
    "member not found": "eligibility",
    "provider was not eligible on dos": "eligibility",
    "ins termed": "eligibility",
}
fwd[target] = fwd[target].map(category_map)

priority_map = {
    "invalid_modifier": 1, "inclusive_procedure": 2, "lack_of_info": 3,
    "noncovered_benefits": 4, "cpt_icd_mismatch": 5, "invalid_pos": 6,
    "gender_inconsistent": 7, "age_inconsistent": 8,
    "authorization": 9, "eligibility": 10, "payer_rule": 11,
}
fwd["priority"] = fwd[target].map(priority_map)
fwd = fwd.sort_values("priority").drop_duplicates(subset=["claim_no", "CPT"], keep="first")
fwd[target] = fwd[target].fillna("cat")
fwd = fwd[fwd[target] != "cat"]

feature_cols_base = [
    "CPT", "M1", "M2", "M3", "M4",
    "D1", "D2", "D3", "D4",
    "P1", "P2", "P3", "P4",
    "ABK",
    "ABF_1", "ABF_2", "ABF_3", "ABF_4", "ABF_5",
    "ABF_6", "ABF_7", "ABF_8", "ABF_9", "ABF_10", "ABF_11",
    "patient_gender", "patient_age", "PAYER_ID", "POS",
]
fwd = fwd[feature_cols_base + ["claim_no", target]].copy()

code_cols_fwd = [c for c in feature_cols_base if c not in ["patient_gender", "patient_age", "PAYER_ID"]]
fwd[code_cols_fwd] = fwd[code_cols_fwd].fillna("MISSING")

# Feature engineering
fwd = fwd.sort_values(["claim_no", "CPT"])
claim_cpts_fwd = fwd.groupby("claim_no")["CPT"].apply(list)
fwd = fwd.merge(claim_cpts_fwd.rename("all_cpts"), on="claim_no")
fwd["other_cpts"] = fwd.apply(lambda x: [c for c in x["all_cpts"] if c != x["CPT"]], axis=1)
fwd["other_cpt_1"] = fwd["other_cpts"].str[0]
fwd["other_cpt_2"] = fwd["other_cpts"].str[1]
fwd["other_cpt_3"] = fwd["other_cpts"].str[2]
fwd[["other_cpt_1", "other_cpt_2", "other_cpt_3"]] = fwd[["other_cpt_1", "other_cpt_2", "other_cpt_3"]].fillna("NONE")
fwd = fwd.drop(columns=["all_cpts", "other_cpts"])

mod_cols_fwd  = ["M1", "M2", "M3", "M4"]
diag_cols_fwd = ["D1", "D2", "D3", "D4", "ABK",
                 "ABF_1", "ABF_2", "ABF_3", "ABF_4", "ABF_5",
                 "ABF_6", "ABF_7", "ABF_8", "ABF_9", "ABF_10", "ABF_11"]

fwd["modifier_count"]   = (fwd[mod_cols_fwd]  != "MISSING").sum(axis=1)
fwd["diagnosis_count"]  = (fwd[diag_cols_fwd] != "MISSING").sum(axis=1)
fwd["cpt_count"]        = fwd.groupby("claim_no")["CPT"].transform("count")
fwd["unique_cpt_count"] = fwd.groupby("claim_no")["CPT"].transform("nunique")

print("After preprocessing:", fwd.shape)
print(fwd[target].value_counts().to_string())

# ----------------------------------------------------------
# 3. Load saved artefacts
# ----------------------------------------------------------
lookup_tables_fwd = pickle.load(open("probability_lookup_tables.pkl", "rb"))

# ----------------------------------------------------------
# 4. Probability encoding using TRAINING lookup tables only
# ----------------------------------------------------------
prob_cols_fwd = [
    "CPT",
    "other_cpt_1", "other_cpt_2", "other_cpt_3",
    "M1", "M2", "M3", "M4",
    "D1", "D2", "D3", "D4",
    "ABK", "PAYER_ID", "POS",
]

for col in prob_cols_fwd:
    fwd[col] = fwd[col].astype(str)

for col in prob_cols_fwd:
    if col not in lookup_tables_fwd:
        continue
    prob_tbl = lookup_tables_fwd[col]
    for cls in prob_tbl.columns:
        new_col = f"{col}_prob_{cls}"
        fwd[new_col] = fwd[col].map(prob_tbl[cls]).fillna(0)

# ----------------------------------------------------------
# 5. Build X / y and align to hardcoded training columns
# ----------------------------------------------------------
non_feat = prob_cols_fwd + [target, "priority", "claim_no"]
X_fwd = fwd.drop(columns=non_feat, errors="ignore").select_dtypes(include=[np.number])
y_fwd = fwd[target]

print(f"\nX_fwd shape : {X_fwd.shape}  |  samples : {len(y_fwd)}")

X_fwd = X_fwd.reindex(columns=TRAIN_COLS, fill_value=0)

# ----------------------------------------------------------
# 6. Load model and predict
# ----------------------------------------------------------
cb_fwd = CatBoostClassifier()
cb_fwd.load_model("catboost_denial_multiclass.cbm")

cb_pred_fwd = cb_fwd.predict(X_fwd).flatten()

# ----------------------------------------------------------
# 7. Grouped evaluation:
#    invalid_modifier / lack_of_info / authorization → shown as-is
#    everything else → "other"
# ----------------------------------------------------------
KEY_CLASSES = {"invalid_modifier", "lack_of_info", "authorization"}

def group_label(lbl):
    return lbl if lbl in KEY_CLASSES else "other"

y_fwd_grp = y_fwd.map(group_label)
pred_grp  = pd.Series(cb_pred_fwd).map(group_label)

print("\n" + "="*60)
print("FORWARD TEST — January 2026")
print(f"Overall Accuracy (grouped) : {accuracy_score(y_fwd_grp, pred_grp):.4f}")
print("="*60)
print(classification_report(y_fwd_grp, pred_grp, digits=4))

print("Per-class accuracy (grouped):")
for cls in sorted(y_fwd_grp.unique()):
    mask = y_fwd_grp == cls
    if mask.sum() == 0:
        continue
    cls_acc = accuracy_score(y_fwd_grp[mask], pred_grp[mask])
    print(f"  {cls:<30s}  acc={cls_acc:.4f}  support={mask.sum()}")
```
